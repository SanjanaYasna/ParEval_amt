```python
import pandas as pd
import matplotlib.pyplot as plt
import os
import numpy as np

# List of input CSVs
csvs = [
    "/work/pi_mrobson_smith_edu/scratch/generation_hpx/fft/driver/tcmalloc/json_runtime_summary.csv"
]

# Thread counts corresponding to runtime columns
threads = [1, 2, 4, 8, 16, 32, 64]
runtime_cols = [f"runtime_{t}" for t in threads]

# Global style tweaks
plt.rcParams.update({
    "font.size": 11,
    "axes.grid": True,
    "grid.alpha": 0.3,
    "grid.linestyle": "--",
    "axes.spines.top": False,
    "axes.spines.right": False,
})

for csv_path in csvs:
    df = pd.read_csv(csv_path)

    # ---- Parse prompt category from path ----
    path_parts = csv_path.split(os.sep)
    try:
        gen_idx = path_parts.index("generation_hpx")
        prompt_category = path_parts[gen_idx + 1]
    except (ValueError, IndexError):
        prompt_category = "unknown"

    plt.figure(figsize=(7, 4.5))

    # Pastel color cycle
    colors = plt.cm.Pastel1(np.linspace(0, 1, len(df)))

    for (_, row), color in zip(df.iterrows(), colors):
        runtimes = [row[col] for col in runtime_cols]
        plt.plot(
            threads,
            runtimes,
            marker="o",
            linewidth=2,
            markersize=6,
            color=color,
            label=row["model"],
        )

    plt.xlabel("Number of Threads")
    plt.ylabel("Runtime")
    plt.title(f"Runtime Scaling — Prompt Category: {prompt_category}")

    # ---- Force x-axis to only show powers of two present ----
    plt.xticks(threads)
    plt.xscale("log", base=2)

    plt.legend(frameon=False)
    plt.tight_layout()
    plt.show()
```
